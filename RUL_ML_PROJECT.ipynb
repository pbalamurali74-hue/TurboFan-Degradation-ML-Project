```python
import pandas as pd

# Load dataset
df = pd.read_csv("train_FD001.txt", sep=r"\s+", header=None)

# Assign column names
columns = [
    "unit_number", "cycle",
    "setting_1", "setting_2", "setting_3",
    "sensor_1", "sensor_2", "sensor_3", "sensor_4",
    "sensor_5", "sensor_6", "sensor_7", "sensor_8",
    "sensor_9", "sensor_10", "sensor_11", "sensor_12",
    "sensor_13", "sensor_14", "sensor_15", "sensor_16",
    "sensor_17", "sensor_18", "sensor_19", "sensor_20",
    "sensor_21"
]

df.columns = columns

# Create duplicate dataset
new_df = df.copy()

# Calculate median
median = new_df["cycle"].median()

print("Median of cycle is:", median)

# Display summary statistics
print("\nSummary Statistics:")
display(new_df.describe())
```
